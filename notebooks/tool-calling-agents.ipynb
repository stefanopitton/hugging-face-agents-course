```python
model = InferenceClientModel(model_id="meta-llama/Llama-3.3-70B-Instruct")

agent = ToolCallingAgent(
    tools=[WebSearchTool()],
    model=model,
)

agent.run("Search for the best music recommendations for a party at the Wayne's mansion. We need at least 10 songs from different band to play and not a single playlist.")
```
